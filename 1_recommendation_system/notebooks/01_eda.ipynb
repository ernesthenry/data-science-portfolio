# Exploratory Data Analysis: MovieLens Dataset

In this notebook, we analyze the user-item interaction matrix, rating distributions, and sparsity which are critical factors for building a recommendation system.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
plt.style.use('ggplot')
sns.set_palette('husl')

print("EDA Environment Ready")
```

## 1. Data Generation (Simulating MovieLens 100k)

Since we are in a development environment, we will generate a synthetic dataset that mirrors the characteristics of MovieLens 100k.

```python
np.random.seed(42)
n_users = 943
n_movies = 1682
n_ratings = 100000

users = np.random.randint(1, n_users + 1, n_ratings)
movies = np.random.randint(1, n_movies + 1, n_ratings)
ratings = np.random.choice([1, 2, 3, 4, 5], n_ratings, p=[0.1, 0.15, 0.3, 0.3, 0.15])
timestamps = pd.to_datetime('2023-01-01') + pd.to_timedelta(np.random.randint(0, 365, n_ratings), unit='D')

df = pd.DataFrame({
    'user_id': users,
    'movie_id': movies,
    'rating': ratings,
    'timestamp': timestamps
})

print(f"Dataset Shape: {df.shape}")
print(df.head())
```

## 2. Rating Distribution

We visualize the frequency of each rating to understand user behavior.

```python
plt.figure(figsize=(10, 6))
sns.countplot(x='rating', data=df)
plt.title('Distribution of Movie Ratings')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.show()
```

## 3. Most Popular Movies

Identifying items with the most interactions to establish a 'Most Popular' baseline.

```python
movie_stats = df.groupby('movie_id').agg({'rating': ['count', 'mean']})
movie_stats.columns = ['rating_count', 'avg_rating']

top_10_popular = movie_stats.sort_values(by='rating_count', ascending=False).head(10)
print("Top 10 Most Rated Movies:")
print(top_10_popular)

plt.figure(figsize=(12, 6))
sns.scatterplot(data=movie_stats, x='rating_count', y='avg_rating', alpha=0.5)
plt.title('Movie Rating Count vs Average Rating')
plt.show()
```

## 4. Sparsity Analysis

Sparsity is a major challenge in collaborative filtering.

```python
n_possible_ratings = n_users * n_movies
n_actual_ratings = len(df)
sparsity = 1 - (n_actual_ratings / n_possible_ratings)

print(f"Number of Users: {n_users}")
print(f"Number of Movies: {n_movies}")
print(f"Matrix Sparsity: {sparsity * 100:.2f}%")
```

## 5. User Interaction Density

How many movies does each user rate on average?

```python
user_counts = df.groupby('user_id').size()
plt.figure(figsize=(10, 6))
sns.histplot(user_counts, bins=50, kde=True)
plt.title('Distribution of Ratings per User')
plt.xlabel('Number of Ratings')
plt.show()
```
