# Project 5: Time Series - Exploratory Data Analysis

Understanding the underlying patterns of retail sales data, including seasonality, trend, and autocorrelation.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

plt.style.use('bmh')
print("TS EDA Ready")
```

## 1. Seasonal Decomposition

Visualizing Trend, Seasonality, and Residuals.

```python
dates = pd.date_range('2021-01-01', periods=730, freq='D')
y = 100 + 0.1 * np.arange(730) + 20 * np.sin(2 * np.pi * np.arange(730) / 365) + np.random.normal(0, 5, 730)
ts = pd.Series(y, index=dates)

result = seasonal_decompose(ts, model='additive')
result.plot()
plt.show()
```

## 2. Autocorrelation (ACF) and Partial Autocorrelation (PACF)

Identifying the order of moving average (MA) and autoregressive (AR) processes.

```python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8))
plot_acf(ts, ax=ax1, lags=50)
plot_pacf(ts, ax=ax2, lags=50)
plt.tight_layout()
plt.show()
```
