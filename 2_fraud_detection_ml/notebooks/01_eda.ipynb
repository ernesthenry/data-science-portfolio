# Multi-Stage Fraud Detection Analysis

Fraud detection requires handling extreme class imbalance and identifying subtle behavioral anomalies.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('fivethirtyeight')
print("Fraud Analysis Environment Ready")
```

## 1. Class Imbalance Visualization

In financial transactions, fraud is a 'needle in a haystack'.

```python
n_samples = 10000
y = np.random.choice([0, 1], n_samples, p=[0.99, 0.01])
df = pd.DataFrame({'is_fraud': y})

plt.figure(figsize=(8, 5))
df['is_fraud'].value_counts().plot(kind='pie', autopct='%1.1f%%', colors=['#66b3ff','#99ff99'])
plt.title('Fraud vs Legitimate Transactions')
plt.show()
```

## 2. Amount Distribution

Fraudulent transactions often have different spending patterns.

```python
legit_amounts = np.random.exponential(100, 9900)
fraud_amounts = np.random.uniform(500, 2000, 100)

plt.figure(figsize=(12, 6))
sns.kdeplot(legit_amounts, label='Legit', fill=True)
sns.kdeplot(fraud_amounts, label='Fraud', fill=True)
plt.title('Transaction Amount Distribution')
plt.xlabel('Amount')
plt.legend()
plt.show()
```

## 3. Correlation with Behavioral Features

Analyzing how features like 'hour of day' or 'device type' correlate with fraud.

```python
hours = np.random.randint(0, 24, 10000)
fraud_hours = np.random.choice([0, 1, 2, 3, 4], 100) # Fraud often happens late at night

plt.figure(figsize=(10, 6))
sns.histplot(hours, label='Full Dataset', color='gray', alpha=0.3, bins=24)
sns.histplot(fraud_hours, label='Fraudulent', color='red', bins=24)
plt.title('Temporal Analysis of Fraud')
plt.xlabel('Hour of Day')
plt.legend()
plt.show()
```
