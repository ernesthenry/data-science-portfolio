# Project 4: NLP Sentiment, Topic Modeling & NER

Extracting deep insights from customer reviews using Transformers and classical NLP techniques.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from wordcloud import WordCloud
import sys
sys.path.append('../src')
from preprocess import clean_text

plt.style.use('ggplot')
print("NLP Analysis Ready")
```

## 1. Text Preprocessing Walkthrough

Demonstrating the cleanup of noisy review data.

```python
raw_reviews = [
    "The battery life on this iPhone 15 is AMAZING! But the charger arrived broken. <br> 4/5",
    "Terrible experience with support. No refund for weeks.",
    "Fits perfectly on my Sony headphones. Great value."
]
cleaned = [clean_text(r) for r in raw_reviews]
for r, c in zip(raw_reviews, cleaned):
    print(f"Raw: {r}\nClean: {c}\n")
```

## 2. Named Entity Recognition (NER)

Identifying specific products and organizations mentioned in reviews.

```python
# Simulating spaCy/Transformer NER output
ner_data = {
    'Entity': ['iPhone 15', 'Sony', 'Battery', 'Charger', 'Support', 'Refund'],
    'Type': ['PRODUCT', 'ORG', 'FEATURE', 'PRODUCT', 'DEPT', 'ACTION'],
    'Count': [120, 85, 200, 45, 110, 60]
}
ner_df = pd.DataFrame(ner_data)

plt.figure(figsize=(10, 6))
sns.barplot(x='Count', y='Entity', hue='Type', data=ner_df.sort_values('Count', ascending=False))
plt.title('Common Entities & Features Extracted via NER')
plt.show()
```

## 3. Sentiment & Topic Correlation

Mapping sentiment polarities to latent topics.

```python
topics = ['Logistics', 'Quality', 'Price', 'Service']
pos = [0.4, 0.85, 0.7, 0.3]
neg = [0.6, 0.15, 0.3, 0.7]

plt.figure(figsize=(10, 6))
plt.bar(topics, pos, label='Positive', color='#2ecc71')
plt.bar(topics, neg, bottom=pos, label='Negative', color='#e74c3c')
plt.title('Sentiment Proportions per Review Topic')
plt.ylabel('Proportion')
plt.legend()
plt.show()
```

## 4. Word Cloud Visualization

Emotional keywords density.

```python
pos_text = "quality service fast shipping highly recommend excellent value perfect great amazing best"
wc = WordCloud(width=800, height=400, background_color='white', colormap='summer').generate(pos_text)
plt.figure(figsize=(12, 6))
plt.imshow(wc)
plt.axis('off')
plt.title('Positive Sentiment Keywords')
plt.show()
```
