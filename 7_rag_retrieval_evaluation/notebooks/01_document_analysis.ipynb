# Project 7: RAG - Document & Embedding Analysis

Analyzing the knowledge base structure, document lengths, and embedding distributions to optimize retrieval parameters like chunk size and overlap.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('ggplot')
print("RAG Data Analysis Ready")
```

## 1. Document Length Distribution

Understanding the raw text statistics.

```python
doc_lengths = np.random.lognormal(mean=6, sigma=1, size=1000)
plt.figure(figsize=(10, 6))
sns.histplot(doc_lengths, bins=50, kde=True)
plt.title('Distribution of Document Lengths (Tokens)')
plt.xlabel('Token Count')
plt.show()
```

## 2. Similarity Search Analysis

Visualizing the cosine similarity of query-document pairs in the vector space.

```python
similarities = np.random.uniform(0.6, 0.95, 100)
plt.figure(figsize=(10, 6))
sns.boxplot(x=similarities, color='skyblue')
plt.title('Distribution of Retrieval Similarity Scores (Top-K)')
plt.xlabel('Cosine Similarity')
plt.show()
```
