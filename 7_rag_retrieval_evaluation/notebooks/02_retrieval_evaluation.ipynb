## 1. Vector Search Simulation

Adding documents to the index and performing semantic searches.

```python
store = VectorStore()
kb = [
    "The transformer architecture uses self-attention mechanisms.",
    "BERT is a bidirectional encoder transformer model.",
    "RAG combines retrieval with generative LLMs.",
    "Vector databases index embeddings for fast retrieval."
]
store.add_documents(kb)

query = "Explain BERT architecture."
results = store.search(query, k=2)
print(f"Top Results for '{query}':")
for i, res in enumerate(results):
    print(f"{i+1}. {res}")
```

## 2. RAG Context Augmentation Trace

Visualizing how the context is injected into the LLM prompt.

```python
rag = RAGSystem(store)
response = rag.query(query)
print(f"Final RAG Response: {response}")
```

## 3. Comprehensive Evaluation: Retrieval Stability

Measuring MRR across different chunk sizes (simulated).

```python
chunk_sizes = [128, 256, 512, 1024]
mrr_scores = [0.82, 0.88, 0.85, 0.78]
hit_rate_5 = [0.89, 0.94, 0.92, 0.85]

plt.figure(figsize=(10, 6))
plt.plot(chunk_sizes, mrr_scores, marker='o', label='MRR')
plt.plot(chunk_sizes, hit_rate_5, marker='s', label='Hit Rate@5')
plt.title('RAG Sensitivity: Chunk Size vs Retrieval Performance')
plt.xlabel('Chunk Size (Tokens)')
plt.ylabel('Score')
plt.legend()
plt.show()
```
