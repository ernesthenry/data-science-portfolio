## 2. Tracking Server Dashboard (Mock)

Visualizing parameter vs metric correlations from multiple runs.

```python
runs = pd.DataFrame({
    'run_id': ['a1', 'b2', 'c3', 'd4'],
    'learning_rate': [0.01, 0.001, 0.05, 0.1],
    'accuracy': [0.88, 0.92, 0.82, 0.75],
    'status': ['FINISHED', 'FINISHED', 'FAILED', 'FINISHED']
})

plt.figure(figsize=(10, 6))
plt.scatter(runs['learning_rate'], runs['accuracy'], s=200, c='green', alpha=0.6)
plt.xscale('log')
plt.title('MLOps: Parameter Hyperparameter Search Visualization')
plt.xlabel('Learning Rate (Log Scale)')
plt.ylabel('Test Accuracy')
plt.grid(True)
plt.show()
```
